```python
import ase
from ase.build import fcc100

import MDAnalysis as mda
import solvate
```

```python
gold_slab = fcc100('Au', size=(4,4,1), vacuum=0, )
neon_slab = fcc100('Ne', size=(4,4,1), vacuum=0, )

gold_slab.translate([0,0,0.5])
neon_slab.translate([0,0,0.5])


neon_slab.translate([0,0,15])

slab = gold_slab + neon_slab

slab.translate([0,0,10])

slab.cell[2][2] = slab.positions[:,2].max() + 10

slab.write('slab.gro')

u = mda.Universe('slab.gro')


u = solvate.InsertPlanar(u, solvate.models.spce(), n=8, zmin=12, zmax=20)

u.atoms.write('solvated_slab.gro')
u.atoms.write('solvated_slab.xyz')
```
